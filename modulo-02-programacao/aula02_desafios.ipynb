```python
# desafio 1: variaveis
# pegue um nome
# cidade
# ano de nascimento
# imprima uma frase com essas variaveis

print("\n desafio 1: variáveis")
nome = input("Digite o seu nome: ")
cidade = input("Digite a sua cidade: ")
nascimento = input("O ano que você nasceu: ")
print(f"\nOlá, {nome}! Você é de {cidade} e nasceu em {nascimento}.")
```

```python
# desafio 2: condicoes
# desenvolva um programa que verifique a idade de uma pessoa se maior de 18 anos
# para determinar se ela pode entrar em um brinquedo de parque.

print("\n desafio 2: condicionais")
idade = int(input("Digite sua idade: "))
if idade >= 18:
    print("Você pode brincar na montanha-russa!")
elif idade >= 15:
    print("Você pode brincar acompanhado de um responsável.")
else:
    print("Você ainda não pode acessar esse brinquedo.")
```

```python
# desafio 3: tabuada
print("\n desafio 3: tabuada")
numero = int(input("Digite um número para ver a tabuada: "))
print()
for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")
```
